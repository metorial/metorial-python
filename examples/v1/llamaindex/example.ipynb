# Metorial + LlamaIndex Example

This notebook demonstrates how to use Metorial tools with LlamaIndex agents.

```python
# Install dependencies
%pip install metorial llama-index llama-index-llms-openai python-dotenv
```

```python
import os

from dotenv import load_dotenv

load_dotenv()

assert os.getenv("METORIAL_API_KEY"), "Set METORIAL_API_KEY"
assert os.getenv("OPENAI_API_KEY"), "Set OPENAI_API_KEY"
assert os.getenv("EXA_DEPLOYMENT_ID"), "Set EXA_DEPLOYMENT_ID"
```

```python
from llama_index.core.agent import ReActAgent
from llama_index.llms.openai import OpenAI

from metorial import Metorial
from metorial.integrations.llamaindex import create_llamaindex_tools
```

```python
metorial = Metorial(api_key=os.getenv("METORIAL_API_KEY"))
```

```python
async def run_agent(query: str):
    async with metorial.provider_session(
        provider="openai",
        server_deployments=[os.getenv("EXA_DEPLOYMENT_ID")],
    ) as session:
        tools = create_llamaindex_tools(session)

        print(f"Available tools: {[t.metadata.name for t in tools]}")

        llm = OpenAI(model="gpt-4o")
        agent = ReActAgent.from_tools(tools, llm=llm, verbose=True)

        response = agent.chat(query)
        return response
```

```python
result = await run_agent("Search for the latest Python 3.13 features")
print(result)
```
