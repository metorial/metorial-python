# Metorial + PydanticAI Example

This notebook demonstrates how to use Metorial tools with PydanticAI agents.

```python
# Install dependencies
%pip install metorial pydantic-ai python-dotenv
```

```python
import os

from dotenv import load_dotenv

load_dotenv()

# Make sure these environment variables are set
assert os.getenv("METORIAL_API_KEY"), "Set METORIAL_API_KEY"
assert os.getenv("OPENAI_API_KEY"), "Set OPENAI_API_KEY"
assert os.getenv("EXA_DEPLOYMENT_ID"), "Set EXA_DEPLOYMENT_ID"
```

```python
from pydantic_ai import Agent

from metorial import Metorial
from metorial.integrations.pydantic_ai import register_metorial_tools
```

```python
metorial = Metorial(api_key=os.getenv("METORIAL_API_KEY"))
```

```python
async def run_agent(query: str):
  async with metorial.provider_session(
    provider="openai",
    server_deployments=[os.getenv("EXA_DEPLOYMENT_ID")],
  ) as session:
    agent = Agent(
      "openai:gpt-4o",
      system_prompt="You are a helpful research assistant.",
    )

    # Register Metorial tools with the agent
    register_metorial_tools(agent, session)

    print("Tools registered with agent")

    result = await agent.run(query)
    return result.data
```

```python
result = await run_agent("Search for the latest Python 3.13 features")
print(result)
```
